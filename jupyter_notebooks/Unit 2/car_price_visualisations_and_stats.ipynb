```python
# Columns to exclude from distribution plots
exclude_cols = ["car_ID", "CarName", "CarModel_Standardised", "doornumber", "symboling"]

# Select relevant numeric columns
numeric_cols = [
    col
    for col in df.select_dtypes(include="number").columns
    if col not in exclude_cols
]

# Disable KDE for categorical or specific column list
no_kde_cols = ["drivewheel", "enginelocation"]

for col in numeric_cols:
    fig, axes = plt.subplots(
        nrows=2,
        ncols=1,
        figsize=(6, 4),
        sharex=True,
        gridspec_kw={"height_ratios": (0.15, 0.85)},
    )

    # Box plot on top
    sns.boxplot(data=df, x=col, ax=axes[0], color="skyblue")
    axes[0].set(xlabel="")  # Remove redundant x-label from top plot

    # Determine KDE setting based on feature name
    use_kde = col not in no_kde_cols

    # Histogram on bottom
    sns.histplot(data=df, x=col, kde=use_kde, ax=axes[1], color="steelblue")

    fig.suptitle(f"{col} Distribution - Boxplot and Histogram", fontsize=12)
    plt.tight_layout()
    plt.show()
    print("\n")
```

```python
categorical_cols = ["drivewheel", "symboling"]

for col in categorical_cols:
    plt.figure(figsize=(6, 4))

    # Create count plot with bar annotations
    ax = sns.countplot(
        data=df,
        x=col,
        order=df[col].value_counts().index,  # Order bars by frequency
        palette="viridis",
        hue=col,
        legend=False
    )

    # Add count labels above each bar
    for container in ax.containers:
        ax.bar_label(container, padding=3)

    # Expand top y-limit by 10% to prevent label clipping
    ax.margins(y=0.1)

    plt.title(f"Count Distribution of {col.capitalize()}", fontsize=12)
    plt.xlabel(col.capitalize(), fontsize=10)
    plt.ylabel("Count", fontsize=10)
    plt.tight_layout()
    plt.show()
    print("\n")
```

```python
# -------------------------------------------------------------------------
# 14. Correlation of numerical variables 
# -------------------------------------------------------------------------
# Select numeric features for Pearson correlation ranking
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()

# Extract top correlated features with price
top_corr = corr_matrix['price'].sort_values(ascending=False)

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix.loc[top_corr.index, top_corr.index], 
    annot=True, fmt=".2f", cmap="coolwarm", cbar=True, linewidths=0.5
)
plt.title("Correlation Matrix Ranked by Target Variable (Price)", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("Top Numerical Correlates with Price:\n", top_corr)
```

```python
import scipy.stats as stats

#  Pearson Correlation for Numerical Features against Price
num_cols = df.select_dtypes(include=[np.number]).columns.drop(['car_ID', 'price'])
corr_results = []

for col in num_cols:
    corr, p_val = stats.pearsonr(df[col], df['price'])
    corr_results.append({'Feature': col, 'Pearson_r': corr, 'p_value': p_val})

corr_df = pd.DataFrame(corr_results).sort_values(by='Pearson_r', ascending=False)
print("--- Numerical Feature Correlation with Price ---")
print(corr_df)
```

```python
#Check skewness and kurtosis of data before performing ANOVA

# 1. Select numeric columns and drop car_ID
df_numeric = df.drop(columns=['car_ID'], errors='ignore').select_dtypes(include=['number'])

# 2. Build summary DataFrame with only Skewness and Kurtosis
summary_stats = pd.DataFrame({
    'Skewness': df_numeric.skew(),
    'Kurtosis': df_numeric.kurtosis()
}).round(2)

summary_stats
```

```python
# 1. Log Transformation for price and enginesize (using log1p handles zeros safely if they exist)
df['log_price'] = round(np.log1p(df['price']),3)
df['log_enginesize'] = round(np.log1p(df['enginesize']),3)

# 2. Square Root Transformation for wheelbase
df['sqrt_wheelbase'] = round(np.sqrt(df['wheelbase']),3)
```

```python
# ANOVA Test for Categorical Features against Price
cat_cols = ['carbody', 'drivewheel', 'enginelocation', 'enginetype', 'fuelsystem', 'fueltype', 'aspiration', 'manufacturer']
anova_results = []

for col in cat_cols:
    groups = [group['log_price'].values for name, group in df.groupby(col)]
    f_stat, p_val = stats.f_oneway(*groups)
    anova_results.append({'Feature': col, 'F_Statistic': f_stat, 'p_value': p_val})

anova_df = pd.DataFrame(anova_results).sort_values(by='F_Statistic', ascending=False)
print("\n--- Categorical Feature ANOVA Test with Price ---")
print(anova_df)
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Select initial numerical candidate features
candidate_num_features = ['curbweight', 'log_enginesize', 'horsepower', 'carwidth', 'boreratio', 'citympg']

X_vif = df[candidate_num_features].dropna()
vif_data = pd.DataFrame()
vif_data["Feature"] = X_vif.columns
vif_data["VIF"] = [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]

print("--- Variance Inflation Factor (VIF) ---")
print(vif_data.sort_values(by="VIF", ascending=False))
```

```python
# -------------------------------------------------------------------------
# Initial feature selection 
# -------------------------------------------------------------------------
# Define Selected Numerical and Categorical Predictors
selected_num = ['curbweight', 'enginesize', 'horsepower', 'citympg']

X_num = df[selected_num].dropna()

#Perform VIF testing on the remaining numerical variables
vif_data_new = pd.DataFrame()
vif_data_new["Feature"] = X_num.columns
vif_data_new["VIF"] = [variance_inflation_factor(X_num.values, i) for i in range(X_num.shape[1])]

print("--- Variance Inflation Factor (VIF) ---")
print(vif_data_new.sort_values(by="VIF", ascending=False))
```

```python
# Testing reduced subset: curbweight, enginesize, citympg
X_vif_clean = df[['curbweight', 'enginesize', 'citympg']].dropna()

vif_data_clean = pd.DataFrame()
vif_data_clean["Feature"] = X_vif_clean.columns
vif_data_clean["VIF"] = [variance_inflation_factor(X_vif_clean.values, i) for i in range(X_vif_clean.shape[1])]

print("--- Updated VIF Output ---")
print(vif_data_clean.sort_values(by="VIF", ascending=False))
```

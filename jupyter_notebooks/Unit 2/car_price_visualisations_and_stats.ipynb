```python
# Columns to exclude from distribution plots
exclude_cols = ["car_ID", "CarName", "CarModel_Standardised", "doornumber", "symboling"]

# Select relevant numeric columns
numeric_cols = [
    col
    for col in df.select_dtypes(include="number").columns
    if col not in exclude_cols
]

# Disable KDE for categorical or specific column list
no_kde_cols = ["drivewheel", "enginelocation"]

for col in numeric_cols:
    fig, axes = plt.subplots(
        nrows=2,
        ncols=1,
        figsize=(6, 4),
        sharex=True,
        gridspec_kw={"height_ratios": (0.15, 0.85)},
    )

    # Box plot on top
    sns.boxplot(data=df, x=col, ax=axes[0], color="skyblue")
    axes[0].set(xlabel="")  # Remove redundant x-label from top plot

    # Determine KDE setting based on feature name
    use_kde = col not in no_kde_cols

    # Histogram on bottom
    sns.histplot(data=df, x=col, kde=use_kde, ax=axes[1], color="steelblue")

    fig.suptitle(f"{col} Distribution - Boxplot and Histogram", fontsize=12)
    plt.tight_layout()
    plt.show()
    print("\n")
```

```python
categorical_cols = ["drivewheel", "symboling"]

for col in categorical_cols:
    plt.figure(figsize=(6, 4))

    # Create count plot with bar annotations
    ax = sns.countplot(
        data=df,
        x=col,
        order=df[col].value_counts().index,  # Order bars by frequency
        palette="viridis",
    )

    # Add count labels above each bar
    for container in ax.containers:
        ax.bar_label(container, padding=3)

    # Expand top y-limit by 10% to prevent label clipping
    ax.margins(y=0.1)

    plt.title(f"Count Distribution of {col.capitalize()}", fontsize=12)
    plt.xlabel(col.capitalize(), fontsize=10)
    plt.ylabel("Count", fontsize=10)
    plt.tight_layout()
    plt.show()
    print("\n")
```

```python
# -------------------------------------------------------------------------
# 1. Body Style vs Price (Sorted by Median Price)
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 4))
body_order = df.groupby('carbody')['price'].median().sort_values(ascending=False).index

sns.boxplot(
    data=df, x='carbody', y='price', order=body_order,
    palette='Set3', hue='carbody', legend=False
)
sns.stripplot(
    data=df, x='carbody', y='price', order=body_order,
    color='black', alpha=0.3, jitter=0.2, size=5
)
plt.title('1. Car Body Style vs Price (Sorted by Median Tier)', fontsize=12, fontweight='bold')
plt.xlabel('Body Style', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------------------
# 2. Compression Ratio vs Price (Colored by Fuel Type)
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 4))
sns.scatterplot(
    data=df, x='compressionratio', y='price', hue='fueltype',
    style='fueltype', palette={'gas': '#e74c3c', 'diesel': '#27ae60'}, s=80
)
plt.title('2. Compression Ratio vs Price (Bimodal Split by Fuel Type)', fontsize=12, fontweight='bold')
plt.xlabel('Compression Ratio', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# 3. Engine Type vs Price (Grouped & Sorted)
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 5))
engine_order = df.groupby('enginetype')['price'].median().sort_values(ascending=False).index

sns.boxplot(
    data=df, x='enginetype', y='price', order=engine_order,
    palette='Pastel1', hue='enginetype', legend=False
)
plt.title('3. Engine Type vs Price (Sorted by Median)', fontsize=12, fontweight='bold')
plt.xlabel('Engine Type', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# 4. Bore Ratio vs Horsepower (Engine Efficiency)
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 5))
sns.regplot(
    data=df, x='boreratio', y='price',
    scatter_kws={'alpha': 0.6, 'color': '#8e44ad'},
    line_kws={'color': '#2c3e50', 'linewidth': 2}
)
plt.title('4. Bore Ratio vs Price', fontsize=12, fontweight='bold')
plt.xlabel('Bore Ratio', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# 5. Peak RPM vs Horsepower (Performance Profile)
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 4))
sns.scatterplot(
    data=df, x='peakrpm', y='horsepower', hue='price',
    palette='viridis', size='price', sizes=(30, 150)
)
plt.title('5. Peak RPM vs Horsepower (Sized & Colored by Price)', fontsize=12, fontweight='bold')
plt.xlabel('Peak RPM', fontsize=10)
plt.ylabel('Horsepower', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------------------
# 6. Combined Dimensional Feature: Car Volume vs Price
# -------------------------------------------------------------------------
df['car_volume'] = df['carlength'] * df['carwidth'] * df['carheight']

plt.figure(figsize=(6, 5))
sns.regplot(
    data=df, x='car_volume', y='price',
    color='#d35400', scatter_kws={'alpha': 0.6}, line_kws={'color': '#2980b9'}
)
plt.title('6. Total Vehicle Volume (Length × Width × Height) vs Price', fontsize=12, fontweight='bold')
plt.xlabel('Car Volume (cubic inches)', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------------------
# 7. Non-Linear Efficiency Curve (City MPG vs Price)
# -------------------------------------------------------------------------
plt.figure(figsize=(7, 5))
sns.regplot(
    data=df, x='citympg', y='price',
    order=2, color='#2b5c8f', scatter_kws={'alpha':0.6}, line_kws={'color':'#e74c3c'}
)
plt.title('A. City MPG vs Price (Non-Linear Quadratic Fit)', fontsize=12, fontweight='bold')
plt.xlabel('City MPG', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.ylim(0, None)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------------------
# 8. Engine Size vs Price
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 4))
sns.regplot(
    data=df, x='enginesize', y='price',
    color='#27ae60', scatter_kws={'alpha':0.6}, line_kws={'color':'#16a085'}
)
plt.title('B. Engine Size vs Price (Primary Physical Driver)', fontsize=12, fontweight='bold')
plt.xlabel('Engine Size', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------------------
# 9. Ordinal Cylinder Progression vs Price
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 4))
sns.boxplot(
    data=df, x='cylindernumber', y='price',
    palette='Blues', order=[2, 3, 4, 5, 6, 8, 12], hue='cylindernumber', legend=False
)
plt.title('C. Price Scaling by Cylinder Count (Ordinal)', fontsize=12, fontweight='bold')
plt.xlabel('Number of Cylinders', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------------------
# 10. Drivewheel Tier Hierarchy vs Price
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 5))
sns.boxplot(
    data=df, x='drivewheel', y='price',
    palette='Set2', hue='drivewheel', legend=False
)
plt.title('D. Drivewheel Configuration Premium', fontsize=12, fontweight='bold')
plt.xlabel('Drivewheel Type', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------------------
# 11. Horsepower vs Price by Aspiration
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 5))
sns.scatterplot(
    data=df, x='horsepower', y='price', hue='aspiration',
    style='aspiration', palette='dark:#5A9_r', s=80
)
plt.title('E. Horsepower & Turbocharger Interaction', fontsize=12, fontweight='bold')
plt.xlabel('Horsepower', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------------------
# 11. Curbweight vs Price by Engine Location
# -------------------------------------------------------------------------
plt.figure(figsize=(6, 4))
sns.scatterplot(
    data=df, x='curbweight', y='price', hue='enginelocation',
    palette={'front':'#34495e', 'rear':'#e74c3c'}, s=80
)
plt.title('F. Curbweight vs Price (Rear Engine Premium)', fontsize=12, fontweight='bold')
plt.xlabel('Curbweight (lbs)', fontsize=10)
plt.ylabel('Price ($)', fontsize=10)
plt.tight_layout()
plt.show()
```
